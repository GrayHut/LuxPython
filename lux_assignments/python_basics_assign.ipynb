```python
#Write a calculator function: Takes two numbers and an operation (+, -, *, /); Returns the result

x=float(input("Enter first number:"))
op=input("Enter the mathematical operation [x;/;+;-]: ")
y=float(input("Enter the second number: "))

def calculator(x,y):
    if op=='x':
        output=x*y
        print(output)
    elif op=='/':
        if x!=0 or y!=0:
            output=x/y
        print(output)
    elif op=='+':
        output=x+y
        print(output)
    elif op=='-':
        output=x-y
        print(output)
        
calculator(x,y)
```
