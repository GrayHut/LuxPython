```python
import pandas as pd
import requests
from sqlalchemy import create_engine
```

```python
weather_source = "https://api.openweathermap.org/data/2.5/weather?q=London&appid=684d4cf93b84151161d5a63c4c624672&units=metric"
response = requests.get(weather_source)

response
```

```python
data = response.json()

data
```

```python
data.keys()
```

```python
data['weather'][0]['description']
```

```python
weather_london = {
    'Weather_Description' : data['weather'][0]['description'],
    'Temperature' : data['main']['temp'],
    'Pressure' : data['main']['pressure']
}

weather_london
```

```python
weather_data = []
weather_data.append(weather_london)
weather_data
```

```python
df = pd.DataFrame(weather_data)
df
```
