> DECORATOR
- Is a function that modifies another function / Decorators take function as an argument and extend the functionality of the said function without modifying it.
- Decorator functions accept atleast one parameter...
- Function & Class decorators

> Example 1

```python
import time

def countdown(func):
    def wrapper(*args,**kwargs):
        start_time=time.time()
        result=func(*args,**kwargs)     #execution of the function
        end_time=time.time()
        print(f"The function: {func.__name__!r} takes {end_time-start_time:4f}secs to run")
        return result
    return wrapper

@countdown
def summation():
    x=int(input("Enter an integer: "))
    return f"Cummulative summation is {sum(range(x+1))}"

print(summation())
```

> Example 2

```python
#this program uses decorator concept to run a function x number of times each time it is called...

import functools

def repeat(num_times):
    def repeat_deco(func):
        @functools.wraps(func)      #preserves metadata(docstring, name) of the original function
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                result=func(*args,**kwargs)
            return result
        return wrapper
    return repeat_deco


@repeat(num_times=5)
def greet(name):
    print(f"Hello {name}") 

greet('Owen')
```

> Example 3(Nested decorators)

```python
import functools

def start_end_deco(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
        print("Start")
        result=func(*args,**kwargs)
        print("End")
        return result
    return wrapper

def debug(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
        args_repr=[repr(a) for a in args]
        kwargs_repr=[f"{k}={v!r}" for k,v in kwargs.items()]
        signature=", ".join(args_repr+kwargs_repr)
        print(f"Calling {func.__name__} ({signature})")
        result=func(*args, **kwargs)
        print(f"{func.__name__!r} returns {result!r}")
        return result
    return wrapper

@debug
@start_end_deco
def say_hello(name):
    greeting=f"Hello {name}"
    print(greeting)
    return greeting

say_hello('GrayHut')
```

> Example 4 (Class decorators)

```python
class CountCalls:
    def __init__(self, func):
        self.func=func
        self.num_calls=0

    def __call__(self, *args, **kwargs):
        self.num_calls+=1
        print(f"This function has been executed {self.num_calls} times")
        return self.func(*args,**kwargs)

@CountCalls
def greet():
    print("Say hello")

greet()
greet()
greet()
```
